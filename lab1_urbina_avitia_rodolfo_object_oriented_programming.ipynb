```python
class Mouse:
    def __init__(self, color, material, size, buttons, type):
        self.color = color
        self.material = material
        self.size = size
        self.buttons = buttons
        self.type = type

    def use(self):
        print("The mouse is being used")

    def move(self):
        print("The mouse is moving")

    def smash(self):
        print("The mouse is being smashed against the floor/wall/desk")

    def describe(self):
        print(f"Color: {self.color}, Material: {self.material}, Size: {self.size}, Buttons: {self.buttons}, Type: {self.type}")


# Creating instances with separate arguments for each attribute
mouse1 = Mouse("White", "Plastic", "Small", 4, "Wireless")
mouse2 = Mouse("Black", "Aluminum", "Medium", 6, "Gaming")
mouse3 = Mouse("Red", "Plastic", "Medium", 5, "Wired")
mouse4 = Mouse("Blue", "Metal", "Large", 8, "Gaming")


print(mouse1.material)
print(mouse2.material)
mouse1.describe()
mouse2.describe()

#LAB1_URBINA_AVITIA_RODOLFO_OBJECT_ORIENTED_PROGRAMMING

class BankAccount:
    def __init__(self, holder, balance):
        self.holder = holder
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("The deposit must be greater than zero")
        else:
            self.__balance += amount
            print(f"${amount} deposited successfully")

    def withdraw(self, amount):
        if amount <= 0:
            print("The withdrawal must be greater than zero")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"${amount} withdrawn successfully")

    def check_balance(self):
        print(f"Current balance: ${self.__balance}")


# Creating two instances
account1 = BankAccount("Raúl Pérez", 5000)
account2 = BankAccount("Joel López", 3000)


# Account 1
print("Account holder:", account1.holder)
account1.check_balance()

account1.deposit(1000)
account1.withdraw(500)

account1.check_balance()


# Account 2
print("\nAccount holder:", account2.holder)
account2.check_balance()

account2.deposit(500)
account2.withdraw(1000)

account2.check_balance()
```
